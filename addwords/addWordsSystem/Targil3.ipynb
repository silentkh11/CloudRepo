```python
from firebase import firebase
import time

# Initialize Firebase connection
FBconn = firebase.FirebaseApplication('https://hamdansproject21-default-rtdb.firebaseio.com/', None)

def get_all_records():
    """Retrieve all temperature records"""
    return FBconn.get('/myTest1/', None)

def update_temperature(record_id, new_temp):
    """Update an existing temperature record"""
    return FBconn.put('/myTest1/' + record_id, 'Temp', new_temp)

def delete_record(record_id):
    """Delete a temperature record"""
    return FBconn.delete('/myTest1/', record_id)

def display_records():
    """Display all temperature records"""
    records = get_all_records()
    if records:
        print("\nCurrent Records:")
        for record_id, data in records.items():
            print(f"ID: {record_id} - Temperature: {data['Temp']}°C")
    else:
        print("\nNo records found")

while True:
    print("\nTemperature Tracker Menu:")
    print("1. Add new temperature")
    print("2. Update existing temperature")
    print("3. Delete temperature record")
    print("4. View all records")
    print("5. Exit")

    choice = input("\nEnter your choice (1-5): ")

    if choice == '1':
        # Add new temperature
        try:
            temperature = int(input("What is the temperature? "))
            data_to_upload = {
                'Temp': temperature,
                'timestamp': time.time()  # Adding timestamp for better tracking
            }
            result = FBconn.post('/myTest1/', data_to_upload)
            print(f"Temperature added successfully! Record ID: {result['name']}")
        except ValueError:
            print("Please enter a valid number for temperature")

    elif choice == '2':
        # Update existing temperature
        display_records()
        record_id = input("\nEnter the record ID to update: ")
        try:
            new_temp = int(input("Enter the new temperature: "))
            result = update_temperature(record_id, new_temp)
            if result:
                print("Temperature updated successfully!")
            else:
                print("Failed to update temperature. Check if the ID exists.")
        except ValueError:
            print("Please enter a valid number for temperature")

    elif choice == '3':
        # Delete temperature record
        display_records()
        record_id = input("\nEnter the record ID to delete: ")
        result = delete_record(record_id)
        if result is None:
            print("Record deleted successfully!")
        else:
            print("Failed to delete record. Check if the ID exists.")

    elif choice == '4':
        # View all records
        display_records()

    elif choice == '5':
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

def take_photo(filename='photo.jpg', quality=0.8):
  js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = 'Capture';
      div.appendChild(capture);

      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      // Resize the output to fit the video element.
      google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

      // Wait for Capture to be clicked.
      await new Promise((resolve) => capture.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);
      stream.getVideoTracks()[0].stop();
      div.remove();
      return canvas.toDataURL('image/jpeg', quality);
    }
    ''')
  display(js)
  data = eval_js('takePhoto({})'.format(quality))
  binary = b64decode(data.split(',')[1])
  with open(filename, 'wb') as f:
    f.write(binary)
  return filename

from IPython.display import Image
try:
  filename = take_photo()
  print('Saved to {}'.format(filename))

  # Show the image which was just taken.
  display(Image(filename))
except Exception as err:
  # Errors will be thrown if the user does not have a webcam or if they do not
  # grant the page permission to access it.
  print(str(err))
```

```python
import IPython
import uuid
from google.colab import output

class InvokeButton(object):
  def __init__(self, title, callback):
    self._title = title
    self._callback = callback

  def _repr_html_(self):
    callback_id = 'button-' + str(uuid.uuid4())
    output.register_callback(callback_id, self._callback)

    template = """<button id="{callback_id}">{title}</button>
        <script>
          document.querySelector("#{callback_id}").onclick = (e) => {{
            google.colab.kernel.invokeFunction('{callback_id}', [], {{}})
            e.preventDefault();
          }};
        </script>"""
    html = template.format(title=self._title, callback_id=callback_id)
    return html

def do_something():
  print('here')

InvokeButton('click me', do_something)

```

```python
from firebase import firebase
import time

# Initialize Firebase connection
FBconn = firebase.FirebaseApplication('https://hamdansproject21-default-rtdb.firebaseio.com/', None)

def get_all_records_words():
    """Retrieve all words"""
    return FBconn.get('/Words/', None)

def get_all_texts():
    """Retrieve all texts"""
    return FBconn.get('/Texts/', None)

def update_word_in_db(record_id, new_word, new_count):
    """Update an existing word record's spelling and base count in Firebase"""
    FBconn.put('/Words/' + record_id, 'Word', new_word)
    return FBconn.put('/Words/' + record_id, 'count', new_count)

def calculate_text_occurrences_for_word(word_to_find):
    """Count how many times a specific word appears in all saved texts"""
    texts_data = get_all_texts()
    count = 0
    if texts_data:
        target_lower = word_to_find.strip().lower()
        for text_id, t_data in texts_data.items():
            if 'Text' in t_data:
                text_content = t_data['Text'].lower()
                words_in_text = text_content.split()
                for w in words_in_text:
                    w_clean = w.strip(".,!?;:()\"'")
                    if w_clean == target_lower:
                        count += 1
    return count

def delete_word_by_value(target_word):
    """Decrease base count by 1. Delete from /Words/ only if total reaches 0. Do not scrub /Texts/."""
    words_data = get_all_records_words()
    target_lower = target_word.strip().lower()
    text_count = calculate_text_occurrences_for_word(target_lower)
    found = False

    if words_data:
        for record_id, data in words_data.items():
            if 'Word' in data and data['Word'].strip().lower() == target_lower:
                found = True
                current_base = int(data.get('count', 0))

                if current_base > 0:
                    current_base -= 1
                    FBconn.put('/Words/' + record_id, 'count', current_base)
                    print(f"Decreased base count of '{target_word}' by 1.")

                    if current_base == 0 and text_count == 0:
                        FBconn.delete('/Words/', record_id)
                        print(f"Word '{target_word}' entirely removed from tracking (total count reached 0).")
                else:
                    # Base count is already 0
                    if text_count == 0:
                        FBconn.delete('/Words/', record_id)
                        print(f"Word '{target_word}' entirely removed from tracking (total count reached 0).")
                    else:
                        print(f"Cannot delete/decrease '{target_word}' further. It has a minimum count of {text_count} from the saved texts.")
    return found

def calculate_word_counts():
    """Calculate total occurrences: Base counts in /Words/ + all word occurrences in /Texts/"""
    words_data = get_all_records_words()
    texts_data = get_all_texts()
    all_word_counts = {}

    # 1. Start with the base user-defined counts from the /Words/ database
    if words_data:
        for w_id, w_data in words_data.items():
            if 'Word' in w_data:
                word_val = w_data['Word'].strip().lower()
                if word_val:
                    all_word_counts[word_val] = all_word_counts.get(word_val, 0) + int(w_data.get('count', 0))

    # 2. Add occurrences found inside the texts (for both tracked and untracked words)
    if texts_data:
        for text_id, t_data in texts_data.items():
            if 'Text' in t_data:
                text_content = t_data['Text'].lower()
                words_in_text = text_content.split()
                for w in words_in_text:
                    w_clean = w.strip(".,!?;:()\"'")
                    if w_clean:
                        # Increment count for any word found in the text
                        all_word_counts[w_clean] = all_word_counts.get(w_clean, 0) + 1
    return all_word_counts

def display_words_with_counts():
    """Display tracked words along with their calculated totals"""
    counts = calculate_word_counts()
    words_data = get_all_records_words()
    if words_data:
        print("\nCurrent Tracked Words and Occurrences:")
        tracked_uniques = sorted(list(set(w_data['Word'].strip() for w_data in words_data.values() if 'Word' in w_data)))
        for unique_word in tracked_uniques:
            word_count = counts.get(unique_word.lower(), 0)
            print(f"Word: {unique_word} - Total Occurrences: {word_count}")
    else:
        print("\nNo tracked words found")

while True:
    print("\nCommon Words Counter Menu:")
    print("1. Add single word")
    print("2. Add text for analysis")
    print("3. Update word count (Base)")
    print("4. Delete/Decrement word count")
    print("5. View all words and occurrence counts")
    print("6. Exit")

    choice = input("\nSelect an option (1 - 6): ")
    if choice == '1':
        # Add new word
        try:
            word = input("What is the word? ").strip()
            if word:
                word_lower = word.lower()
                words_data = get_all_records_words() or {}

                # Find existing database records matching this word
                matching_records = [
                    (r_id, r_data) for r_id, r_data in words_data.items()
                    if 'Word' in r_data and r_data['Word'].strip().lower() == word_lower
                ]

                if matching_records:
                    # Increment count on existing records
                    for r_id, r_data in matching_records:
                        current_count = int(r_data.get('count', 1))
                        new_count = current_count + 1
                        FBconn.put('/Words/' + r_id, 'count', new_count)
                    print(f"Word '{word}' already exists. Its base count in the DB was incremented by 1.")
                else:
                    # Word does not exist, initialize with base count of 1
                    data_to_upload = {
                        'Word': word,
                        'count': 1,
                        'timestamp': time.time()
                    }
                    FBconn.post('/Words/', data_to_upload)
                    print(f"Word '{word}' added successfully with an initial base count of 1.")
            else:
                print("Word cannot be empty.")
        except Exception as e:
            print(f"Error adding/updating word: {e}")

    elif choice == '2':
        # Add text for analysis
        try:
            text = input("What is the text? ").strip()
            if text:
                # 1. Upload text to database
                data_to_upload = {
                    'Text': text,
                    'timestamp': time.time()
                }
                result = FBconn.post('/Texts/', data_to_upload)
                print("Text added successfully!")

                # 2. Extract words from the text and register them into '/Words/' if they aren't already tracked
                words_data = get_all_records_words() or {}
                tracked_words_set = {
                    w_data['Word'].strip().lower() for w_data in words_data.values() if 'Word' in w_data
                }

                words_in_text = text.lower().split()
                new_words_registered = 0
                for w in words_in_text:
                    w_clean = w.strip(".,!?;:()\"'")
                    # If it's a valid word and not already tracked, save it to the '/Words/' node with an initial base count of 0
                    if w_clean and (w_clean not in tracked_words_set):
                        new_word_payload = {
                            'Word': w_clean,
                            'count': 0,
                            'timestamp': time.time()
                        }
                        FBconn.post('/Words/', new_word_payload)
                        tracked_words_set.add(w_clean) # Avoid posting duplicate entries in the same session run
                        new_words_registered += 1

                if new_words_registered > 0:
                    print(f"Automatically registered {new_words_registered} new words from text into tracked list!")
            else:
                print("Text cannot be empty.")
        except Exception as e:
            print(f"Error adding text and parsing words: {e}")

    elif choice == '3':
        # Update base word count only
        words_data = get_all_records_words()
        if not words_data:
            print("\nNo tracked words to update.")
            continue

        display_words_with_counts()
        target_word = input("\nEnter the word you want to update the count for: ").strip()
        matching_ids = [w_id for w_id, w_data in words_data.items() if 'Word' in w_data and w_data['Word'].strip().lower() == target_word.lower()]

        if matching_ids:
            try:
                desired_total = int(input(f"Enter the exact target count you want to display for '{target_word}': "))
            except ValueError:
                print("Invalid count entered. Update cancelled.")
                continue

            # Calculate text occurrences for this word
            text_count = calculate_text_occurrences_for_word(target_word)

            if desired_total < text_count:
                print(f"\n⚠️ Cannot set the total count below {text_count} because it appears {text_count} times in the saved texts.")
                print(f"Correcting the value to the minimum possible: {text_count}.")
                desired_total = text_count

            # Adjusted base count = (desired_total - text_count)
            adjusted_base = desired_total - text_count

            for r_id in matching_ids:
                update_word_in_db(r_id, target_word, adjusted_base)
            print(f"Word count adjusted successfully! It will now display exactly as {desired_total}.")
        else:
            print("Could not find that word in your list.")

    elif choice == '4':
        # Delete/Decrement word
        display_words_with_counts()
        target_word = input("\nEnter the word to delete/decrement: ").strip()
        if target_word:
            success = delete_word_by_value(target_word)
            if not success:
                print(f"Could not find '{target_word}' in your tracked words list.")
        else:
            print("Word input cannot be empty.")

    elif choice == '5':
        # Dynamically calculate total counts and display on-the-fly without modifying base values in DB
        all_word_counts = calculate_word_counts()

        if not all_word_counts:
            print("\nNo words or texts found in the database.")
            continue

        print("\n--- Word Occurrence Summary (Base + Texts) ---")
        output_parts = []
        for unique_word in sorted(all_word_counts.keys()):
            total_count = all_word_counts[unique_word]
            output_parts.append(f"{unique_word}: {total_count}")

        print(", ".join(output_parts))
        print("-" * 46)

    elif choice == '6':
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please try again.")
```
